```python
import time

import py4hw
import py4hw.external.platforms.terasic as terasic
import py4hw.external.wsl.edalize as wsl
```

```python
hw = terasic.DE10_Lite()

hexs = [hw.getOutputHex(i) for i in range(6)]
q = hw.wire('q', 32)
one = hw.wire('one')
              
reset = hw.wire('reset')
second_tick = hw.wire('second_tick')
seconds = hw.wire('seconds', 16)
py4hw.Constant(hw, 'one', 1, one)
py4hw.Constant(hw, 'reset', 0, reset)
py4hw.ModuloCounter(hw, 'mod', reset=reset, inc=one, mod=int(50E6), q=q, carryout=second_tick)
py4hw.Counter(hw, 'counter', reset=reset, inc=second_tick, q=seconds)
bcds = hw.wires('bcd', 6, 4)
py4hw.BinaryToBCDDigits(hw, 'bin2bcd', seconds, bcds)

for i in range(6):
    py4hw.Digit7Segment(hw, f'ss{i}', bcds[i], hexs[i])
```
